```python
import pandas as pd
import numpy as np
import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, InputLayer, Dropout
```

```python
# caminho para os arquivos
path_x = '../archives/entradas_breast.csv'
path_y = '../archives/saidas_breast.csv'

# lendo os arquivos de entrada e saida
X = pd.read_csv(path_x)
y = pd.read_csv(path_y)
```

```python
# criando o classificador e adicionando as camadas
classifier = Sequential()
classifier.add(InputLayer(shape=(30,)))
classifier.add(Dense(units=16, activation='relu', kernel_initializer='random_uniform'))
classifier.add(Dropout(0.2))
classifier.add(Dense(units=16, activation='relu', kernel_initializer='random_uniform'))
classifier.add(Dropout(0.2))
classifier.add(Dense(units=1, activation='sigmoid'))
```

```python
# compilando
classifier.compile(optimizer='adam', loss='binary_crossentropy', metrics=['binary_accuracy'])
```

```python
# treinando o modelo
classifier.fit(X, y, batch_size=10, epochs=100)
```
