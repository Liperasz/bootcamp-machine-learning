```python
import pandas as pd
import tensorflow as tf
import sklearn
import scikeras

from scikeras.wrappers import KerasClassifier
from sklearn.preprocessing import LabelEncoder
from tensorflow.keras import utils as np_utils
from tensorflow.keras import backend as k
from tensorflow.keras.models import Sequential
from sklearn.model_selection import cross_val_score
```

```python
# caminho para o dataset
path = '../archives/iris.csv'

# lendo o csv
df = pd.read_csv(path)

# definindo os atributos previsores e o resultado
X = df.iloc[:, 0:4].values
y = df.iloc[:, 4].values

# transformando a resposta categorica em uma matriz binaria
le = LabelEncoder()
y = le.fit_transform(y)
y = np_utils.to_categorical(y)
```

```python
def create_network():
    k.clear_session()
    # criando a rede neural
    neural_network = Sequential([
        tf.keras.layers.InputLayer(shape=(4,)), # camada de entrada
        tf.keras.layers.Dense(units=4, activation='relu'), # camada densa
        tf.keras.layers.Dense(units=4, activation='relu'), # camada densa
        tf.keras.layers.Dense(units=3, activation='softmax') # camada de saida
    ])
    # compilando
    neural_network.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['categorical_accuracy'])
    return neural_network
```

```python
# criando o modelo com o Keras
neural_network = KerasClassifier(model=create_network, epochs=250, batch_size=10)
```

```python
# fazendo o treinamento com o metodo cruzado
results = cross_val_score(estimator=neural_network, X=X, y=y, cv=10, scoring='accuracy')
```
