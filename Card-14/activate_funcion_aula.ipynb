```python
import numpy as np
```

### Step Funcion

```python
# funcao em degrau (retorna apenas 0 e 1
def step_function(soma):
    if soma >= 1:
        return 1
    return 0

r = step_function(-10)
r
```

### Sigmoid

```python
# função sigmoide (retorna valores entre 0 e 1)
def sigmoid_function(soma):
    return 1 / (1 + np.exp(-soma))

r = sigmoid_function(-10)
r
```

```python
r = sigmoid_function(10)
r
```

### Hyperbolic

```python
# função tanh (retorna valores entre -1 e 1)
def tahn_function(soma):
    return (np.exp(soma) - np.exp(-soma)) / (np.exp(soma) + np.exp(-soma))

r = tahn_function(-10)
r
```

```python
r = tahn_function(10)
r
```
