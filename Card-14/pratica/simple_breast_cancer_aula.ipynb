### Dataset

link: https://archive.ics.uci.edu/dataset/17/breast+cancer+wisconsin+diagnostic

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix
import tensorflow as tf
from tensorflow.keras.models import Sequential
```

## Dataframe

```python
# caminho para o arquivo de entrada
path_x = 'archives/entradas_breast.csv'

# lendo o dataframe
df_x = pd.read_csv(path_x)
# mostrando o dataframe
df_x
```

```python
# caminho para o dataset com os resultados finais (se é ou nao cancer)
path_y = 'archives/saidas_breast.csv'

# lendo e mostrando o dataframe
df_y = pd.read_csv(path_y)
df_y
```

## Train/test Split

```python
# Separando os dados entre treino e teste
X_train, X_test, y_train, y_test = train_test_split(df_x, df_y, test_size=0.25)
```

```python
# vendo a quantidade de linhas e colunas do treino e do teste dos parametros e da resposta
print(X_train.shape)
print(y_train.shape)

print(X_test.shape)
print(y_test.shape)
```

## Criando a rede neural

```python
# criando uma rede neural com o tensor flow camada por camada
neural = Sequential([tf.keras.layers.InputLayer(shape=(30,)), # camada de entrada, com 30 dados de entrada
                     tf.keras.layers.Dense(units=16, activation='relu', kernel_initializer='random_uniform'), # Camada oculta, com 16 neuronios, com a função de ativação relu, e utilizando de pesos aleatórios
                     tf.keras.layers.Dense(units=16, activation='relu', kernel_initializer='random_uniform'), # Criando agora uma outra camada oculta
                     tf.keras.layers.Dense(units=1, activation='sigmoid') # Camada de saída, com uma unica saída, função sigmoid pois a resposta tem de ser binária
                     ])
```

```python
# ver os dados da rede neural
neural.summary()
```

## Configurando a rede neural

```python
# criando agora um optimizador para a rede neural
optimizer = tf.keras.optimizers.Adam(learning_rate=0.001, clipvalue=0.5)
```

```python
# definindo alguns parametros da rede neural
# optimizer é o algoritmo de reajuste dos pesos
# loss é para definidar a função que vai calcular o erro
# metrics são as metricas que serao retornadas com a rede neural
neural.compile(optimizer=optimizer, loss='binary_crossentropy', metrics=['binary_accuracy'])
```

## Treinando

```python
# trainando a rede neural
neural.fit(X_train, y_train, batch_size=10, epochs=100)
```

## Verificando as previsões, acurácia e erro

```python
# fazendo as previsoes de teste
predict = neural.predict(X_test)
predict = predict > 0.5
```

```python
# vendo a precisão do modelo
accuracy_score(y_test, predict)
```

```python
# vendo a matriz de confusão
confusion_matrix(y_test, predict)
```

```python
# vendo o valor do erro e da accuracy com o metodo da rede neural
neural.evaluate(X_test, y_test)
```
