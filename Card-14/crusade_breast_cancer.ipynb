```python
import pandas as pd
import tensorflow as tf
from scikeras.wrappers import KerasClassifier
from sklearn.model_selection import cross_val_score
from tensorflow.keras.models import Sequential
from tensorflow.keras import backend as k
```

```python
# caminho para os arquivos
path_x = 'archives/entradas_breast.csv'
path_y = 'archives/saidas_breast.csv'

X = pd.read_csv(path_x)
y = pd.read_csv(path_y)
```

```python
# função para criar uma rede neural
def create_network():
    # criando uma sessão vazia
    k.clear_session()
    # criando a rede neural
    neural_network = Sequential([
        tf.keras.layers.InputLayer(shape=(30,)), # camada de entrada, com 30 dados de entrada
        tf.keras.layers.Dense(units=16, activation='relu', kernel_initializer='random_uniform'), # Camada oculta, com 16 neuronios, com a função de ativação relu, e utilizando de pesos aleatórios
        tf.keras.layers.Dense(units=16, activation='relu', kernel_initializer='random_uniform'), # Outra camada oculta
        tf.keras.layers.Dense(units=1, activation='sigmoid') # Camada de saída
    ])
    # criando o optimizador
    optimizer = tf.keras.optimizers.Adam(learning_rate=0.001, clipvalue=0.5)
    # configurando a rede neural
    neural_network.compile(optimizer=optimizer, loss='binary_crossentropy', metrics=['binary_accuracy'])
    return neural_network

neural_network = create_network()
neural_network.summary()
```

```python
# fazendo o treinamento com o modelo keras
neural_network = KerasClassifier(model=create_network, epochs=100, batch_size=10)
```

```python
# fazendo o treinamento com o cross validation
results = cross_val_score(estimator=neural_network, X=X, y=y, cv=10, scoring='accuracy')
```
