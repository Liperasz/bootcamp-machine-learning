```python
import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
cv2.__version__
```

```python
import tensorflow
tensorflow.__version__
```

```python
# caminho para o dataset
path = '../fer_images/fer2013/fer2013.csv'

# carregando o dataset
data = pd.read_csv(path)
data.tail()
```

```python
# mostrando o grafio da quantidade de imagens por emoções
plt.figure(figsize=(12,6))
plt.hist(data['emotion'], bins = 30)
plt.title('Imagens x Emoções')
```

```python
# transforma os pixels das imagens em listas
pixels = data['pixels'].tolist()
pixels
```

```python
# definindo as dimensões das imagens
width, height = 48, 48
# lista que vai guardar as faces
faces = []
# quantidade de amostras
sample = 0
# sequência de pixels na imagem
for pixel_sequence in pixels:

    # pegando os pixels de cada imagem
    face = [int(pixel) for pixel in pixel_sequence.split(' ')]
    # transformando em uma matriz numpy
    face = np.array(face).reshape(width, height).astype(np.uint8)
    # adicionando a lista de faces
    faces.append(face)

    if (sample < 10):
        # mostra as primeiras faces
        cv2.imshow(' ', face)
        # pequeno intervalo para que as imagens não travem
        cv2.waitKey(100)
    sample += 1

# fechar as janelas abertas pelo imshow
cv2.destroyAllWindows()
```

```python
faces[:3]
```

```python
print(f"Número de imagens totais no dataset: {str(len(faces))}")
```

```python
# transformando a lista faces em um array do numpy
faces = np.asarray(faces)
faces.shape
```

```python
# adicionando mais uma camada ao final (canal da imagem)
faces = np.expand_dims(faces, -1)
faces.shape
```

```python
# função para normalizar os valores
def normalizate(x):
    x = x.astype('float32')
    x = x / 255.0
    return x
```

```python
# normalizando
faces = normalizate(faces)
faces[0]
```

```python
# transformando os IDs das emoções em arrays do numpy
emotions = pd.get_dummies(data['emotion'], dtype='uint8').values
emotions
```

```python
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Activation, Flatten
from tensorflow.keras.layers import Conv2D, MaxPooling2D, BatchNormalization
from tensorflow.keras.losses import categorical_crossentropy
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.regularizers import l2
from tensorflow.keras.callbacks import ReduceLROnPlateau, EarlyStopping, ModelCheckpoint
from tensorflow.keras.models import load_model
from tensorflow.keras.models import model_from_json
```

```python
# Separando entre treino e teste
X_train, X_test, y_train, y_test = train_test_split(faces, emotions, test_size=0.1, random_state=42)
# separando os dados de validação
X_train, X_val, y_train, y_val = train_test_split(X_train, y_train, test_size=0.1, random_state=41)
```

```python
print(f'Número de imagens no conjunto de treinamento: {len(X_train)}')
print(f'Número de imagens no conjunto de testes: {len(X_test)}')
print(f'Número de imagens no conjunto de validação: {len(X_val)}')
```

```python
# salvando os dados de teste em arquivos
np.save('mod_xtest', X_test)
np.save('mod_ytest', y_test)
```

```python
# numero de filtros da rede neural
num_features = 64
# quantidade de classes
num_labels = 7
# tamanho do lote de registros
batch_size = 64
# epocas (modifiquei de 100 para apenas 5 para que não demorasse tanto, a estivamtiva para as 100 épocas era de 20h para o treinamento)
epochs = 5
# dimensoes das imagens
width, height = 48, 48

# instanciando o modelo
model = Sequential()

# adicionando as camadas

# bloco 1
# convoluções
model.add(Conv2D(num_features, kernel_size=(3,3), activation='relu', input_shape=(width, height, 1), data_format='channels_last', kernel_regularizer=l2(0.01)))
model.add(Conv2D(num_features, kernel_size=(3,3), activation='relu', padding='same'))
# normalização
model.add(BatchNormalization())
# pooling
model.add(MaxPooling2D(pool_size=(2,2), strides=(2,2)))
# dropout (evitar overfiting)
model.add(Dropout(0.5))

# bloco 2
model.add(Conv2D(2 *num_features, kernel_size=(3,3), activation='relu', padding='same'))
model.add(BatchNormalization())
model.add(Conv2D(2 *num_features, kernel_size=(3,3), activation='relu', padding='same'))
model.add(BatchNormalization())
model.add(MaxPooling2D(pool_size=(2,2), strides=(2,2)))
model.add(Dropout(0.5))

# bloco 3
model.add(Conv2D(2 * 2 * num_features, kernel_size=(3,3), activation='relu', padding='same'))
model.add(BatchNormalization())
model.add(Conv2D(2 * 2 * num_features, kernel_size=(3,3), activation='relu', padding='same'))
model.add(BatchNormalization())
model.add(MaxPooling2D(pool_size=(2,2), strides=(2,2)))
model.add(Dropout(0.5))

# bloco 4
model.add(Conv2D(2 * 2 * 2 * num_features, kernel_size=(3,3), activation='relu', padding='same'))
model.add(BatchNormalization())
model.add(Conv2D(2 * 2 * 2 * num_features, kernel_size=(3,3), activation='relu', padding='same'))
model.add(BatchNormalization())
model.add(MaxPooling2D(pool_size=(2,2), strides=(2,2)))
model.add(Dropout(0.5))

# convertendo a matriz em array
model.add(Flatten())

# bloco de camadas densas
model.add(Dense(2*2*2*num_features, activation='relu'))
model.add(Dropout(0.4))
model.add(Dense(2*2*num_features, activation='relu'))
model.add(Dropout(0.4))
model.add(Dense(2*num_features, activation='relu'))
model.add(Dropout(0.4))
model.add(Dense(num_labels, activation='softmax'))

model.summary()
```

```python
# instanciando o otimizador personalizado
adam = Adam(learning_rate= 0.001, beta_1=0.9, beta_2=0.999, epsilon=1e-7)
# compilando o modelo
model.compile(loss = 'categorical_crossentropy', optimizer= adam, metrics=['accuracy'])

# arquivos do modelo
archive_model = 'model_01_expression.h5'
archive_model_json = 'model_01_expression.json'

# Reduz a taxa de aprendizado se o modelo parar de evoluir
lr_reducer = ReduceLROnPlateau(monitor='val_loss', factor=0.9, patience=3, verbose=1)
# Interrompe o treinamento de o modelo parar de aprender
early_stopper = EarlyStopping(monitor='val_loss', min_delta=0, patience=8, verbose=1, mode='auto')
# Salva os pesos em um arquivo de treinamento
checkpointer = ModelCheckpoint(archive_model, monitor='val_loss', verbose=1, save_best_only=True)
```

```python
# salvando a arquitetura do modelo em um json
model_json = model.to_json()
with open(archive_model_json, 'w') as json_file:
    json_file.write(model_json)
```

```python
# Fazendo o treinamento
history = model.fit(np.array(X_train),
                    np.array(y_train),
                    batch_size=batch_size,
                    epochs=epochs,
                    verbose=1,
                    validation_data=(np.array(X_val), np.array(y_val)),
                    shuffle=True,
                    callbacks=[lr_reducer, early_stopper, checkpointer])
```

```python
# historico do treinamento
print(history.history)
```

```python
# Criar grafico com historico do modelo
def plot_history_model(history_model):
    fig, axs = plt.subplots(1,2, figsize=(15,5))

    # grafico da acuracia
    axs[0].plot(range(1, len(history_model.history['accuracy']) + 1), history_model.history['accuracy'], 'r')
    axs[0].plot(range(1, len(history_model.history['val_accuracy']) + 1), history_model.history['val_accuracy'], 'r')
    axs[0].set_title('Acuracia do modelo')
    axs[0].set_ylabel('Acuracia')
    axs[0].set_xlabel('Epoca')
    axs[0].set_xticks(np.arange(1, len(history_model.history['accuracy']) + 1,
                                len(history_model.history['accuracy']) / 10))
    axs[0].legend(['Acuracia do treinamento', 'Acuracia da validacao'], loc='best')

    # grafico da loss
    axs[1].plot(range(1, len(history_model.history['loss']) + 1), history_model.history['loss'], 'r')
    axs[1].plot(range(1, len(history_model.history['val_loss']) + 1), history_model.history['val_loss'], 'r')
    axs[1].set_title('Loss do modelo')
    axs[1].set_ylabel('Loss')
    axs[1].set_xlabel('Epoca')
    axs[1].set_xticks(np.arange(1, len(history_model.history['loss']) + 1,
                                len(history_model.history['loss']) / 10))
    axs[1].legend(['Loss do treinamento', 'Loss da validacao'], loc='best')

    # salva a imagem do grafico
    fig.savefig('history_model_mod01.png')

plot_history_model(history)
```

```python
# vendo as pontuações do modelo
scores = model.evaluate(np.array(X_test), np.array(y_test), batch_size=batch_size)
print(f"Acuracia: {str(scores[1])}")
print(f"Erro: {str(scores[0])}")
```

```python
# lista com as respostas previstas
yp = y_pred.tolist()
# lista das respostas reais
yt = y.tolist()
count = 0

for i in range (len(y)):
    # respostas previstas
    yy = max(yp[i])
    # respostas reais
    yyt = max(yt[i])
    # adicionando as respostas nas listas respectivas
    pred_y.append(yp[i].index(yy))
    true_y.append(yt[i].index(yyt))

    # verificando se acertou
    if (yp[i].index(yy) == yt[i].index(yyt)):
        count+=1

# acuracia
acc = (count / len(y)) * 100
```

```python
from sklearn.metrics import confusion_matrix
```

```python
# carregandos os valores das respostas
y_true = np.load('truey_mod01.npy')
y_pred = np.load('predy_mod01.npy')
```

```python
# criando a matriz de confusão
cm = confusion_matrix(y_true, y_pred)
# expressões faciais
expressions = ['Raiva', 'Nojo', 'Medo', 'Feliz', 'Triste', 'Surpreso', 'Neutro']
title = 'Matriz de confusão'
print(cm)
```

```python
import itertools

# criando a imagem da matriz de confusão
plt.imshow(cm, interpolation='nearest', cmap=plt.cm.Blues)
plt.title(title)
plt.colorbar()
tick_marks = np.arange(len(expressions))
plt.xticks(tick_marks, expressions, rotation=45)
plt.yticks(tick_marks, expressions)
fmt = 'd'
thresh = cm.max() / 2
for i, j in itertools.product(range(cm.shape[0]), range(cm.shape[1])):
    plt.text(j, i, format(cm[i, j], fmt), horizontalalignment='center', color='white' if cm[i,j] > thresh else 'black')

plt.ylabel('Classificacao coreta')
plt.xlabel('Predicao')
plt.savefig('confusion_matrix_mod01.png')
```

```python
# copia da imagem
original = img.copy()
# escala de cinza
gray = cv2.cvtColor(original, cv2.COLOR_BGR2GRAY)
# mostrando a imagem
cv2.imshow(' ', gray)
cv2.waitKey(500)
cv2.destroyAllWindows()
```

```python
# carregando o reconhecedor de faces
face_cascade = cv2.CascadeClassifier('../Material/Material/haarcascade_frontalface_default.xml')
faces = face_cascade.detectMultiScale(gray, 1.1, 3)
```

```python
for (x, y, w, h) in faces:
    # criando o retangulo em volta dos rostos
    cv2.rectangle(original, (x,y), (x+w, y+h), (0,255,0), 1)

    # pegando a area de interesse
    roi_gray = gray[y:y + h, x:x + w]
    # normalizando
    roi_gray = roi_gray.astype('float') / 255.0
    # imagem recortada
    cropped_img = np.expand_dims(np.expand_dims(cv2.resize(roi_gray, (48,48)), -1), 0)
    
    # fazendo a previsão com o modelo
    prediction = loaded_model.predict(cropped_img)[0]

    cv2.putText(original,
                expressions[int(np.argmax(prediction))], (x,y - 10),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.6,
                (0,255,0),
                2,
                cv2.LINE_AA)
    
cv2.imshow(' ', original)
cv2.waitKey(5000)
cv2.destroyAllWindows()
```
