```python
import spacy
from spacy.matcher import Matcher
from spacy.tokens import Span
from spacy.util import filter_spans
```

```python
# carregando o pipeline em portugues
nlp = spacy.load("pt_core_news_sm")
```

```python
# abrindo o arquivo do dom casmurro
with open("domcasmurro.txt", "r") as f:
    text = f.read()

# capítulo 2 até o 4
text = text[5000:10000]
text
```

```python
# criando o objeto doc
doc = nlp(text)
doc[:10]
```

```python
# vendo os primeiros 10 tokens do doc
for token in doc[:10]:
    print(token.text, token.pos_, token.dep_)
```

```python
# vendo as sentenças do doc
for sent in doc.sents:
    print(sent)
```

```python
# pegando a primeira sentença
sentence1 = list(doc.sents)[0]
print(sentence1)
```

```python
# analisando cada token da primeira sentença
for token in sentence1:
    print(token.text, token.pos_, token.lemma_, token.morph)
```

```python
# pegando um token específico e vendo seus atributos
token2 = sentence1[1]
print(token2.text)
print(token2.pos_)
print(token2.dep_)
print(token2.lemma_)
print(token2.ent_iob_)
```

```python
# vendo as entidades nomeadas
for ent in doc.ents:
    print(ent.text, ent.label_)
```

```python
# encontrando nome próprio seguido do verbo (encontrar quem fez tal ação)
matcher = Matcher(nlp.vocab)

# padrão para encontrar com o matcher
pattern = [
    {"POS": "PROPN", "OP": "+"},
    {"POS": "VERB"}
]
# adicionando os padrões no matcher
matcher.add("PROPN_VERB", [pattern], greedy="LONGEST")

# vendo os matches que deu
matches = matcher(doc)
for match in matches[:10]:
    print(match, doc[match[1]:match[2]])
```

```python
# verificando com dois padrões
matcher2 = Matcher(nlp.vocab)

# nome pŕoprio mais ação
pattern_a = [{"POS": "PROPN", "OP": "+"}, {"POS": "VERB"}]
# nome próprio, ação, nome próprio
pattern_b = [{"POS": "PROPN", "OP": "+"}, {"POS": "VERB"}, {"POS": "PROPN", "OP": "+"}]

matcher2.add("PROPN_VERB", [pattern_a], greedy="LONGEST")
matcher2.add("PROPN_VERB_PROPN", [pattern_b], greedy="LONGEST")

matches2 = matcher2(doc)
for match in matches2[:10]:
    # vendo qual regra foi ativada
    print(nlp.vocab[match[0]].text, "->", doc[match[1]:match[2]])
```

```python
# adicionando personagens do livro como entidade customizada ---
nlp2 = spacy.load("pt_core_news_sm")

# adicionando o ruler antes do ner, para ter prioridade
ruler = nlp2.add_pipe("entity_ruler", before="ner")

# personagens principais de dom casmurro
patterns = [
    {"label": "PERSONAGEM", "pattern": "Bentinho"},
    {"label": "PERSONAGEM", "pattern": "Capitú"},
    {"label": "PERSONAGEM", "pattern": "José Dias"},
    {"label": "PERSONAGEM", "pattern": "D. Gloria"},
    {"label": "PERSONAGEM", "pattern": "Tio Cosme"},
    {"label": "PERSONAGEM", "pattern": "Padua"},
]
ruler.add_patterns(patterns)
```

```python
# vendo os pipes do modelo com o ruler adicionado
nlp2.analyze_pipes()
```

```python
# criando doc com o novo pipeline
doc2 = nlp2(text)

# listando entidades, agora com personagem sendo reconhecido reconhecido
for ent in doc2.ents:
    print(ent.text, ent.label_)
```
