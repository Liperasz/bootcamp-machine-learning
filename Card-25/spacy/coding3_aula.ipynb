```python
import spacy
```

```python
# carrega o pipeline do ingles
nlp = spacy.load('en_core_web_sm')
# texto para exemplo
text = 'West Chestertenfieldville was referenced in Mr. Deeds.'
```

```python
# criando um doc
doc = nlp(text)
print(doc)
```

```python
# entidades do doc atualmente
for ent in doc.ents:
    print(ent.text, ent.label_)
```

```python
# instanciando o ruler
ruler = nlp.add_pipe('entity_ruler')
nlp.analyze_pipes()
```

```python
# criando novos padroes para serem reconhecidos
patterns = [
    {'label': 'GPE', 'pattern': 'West Chestertenfieldville'},
    {'label': 'FILM', 'pattern': 'Mr. Deeds'}
]
```

```python
# carregando o modelo em ingles novamente
nlp2 = spacy.load('en_core_web_sm')
```

```python
# adicionando o ruler antes do 'ner', pipe que é responsável justamente por nomear as entidades
ruler = nlp2.add_pipe('entity_ruler', before='ner')
# adiciona o novo padrao
ruler.add_patterns(patterns)
```

```python
# cria um novo doc com o novo padrao
doc2 = nlp2(text)
for ent in doc2.ents:
    print(ent.text, ent.label_)
```

```python
nlp2.analyze_pipes()
```
