```python
import spacy

# texto de exemplo
text = "This is a sample number 555-5555."

# pipeline vazio
nlp = spacy.blank("en")

# cria um ruler e adiciona no pipeline
ruler = nlp.add_pipe("entity_ruler")

# lista os padrões a serem identificados
patterns = [
                {"label": "PHONE_NUMBER", "pattern": [{"SHAPE": "ddd"},
                {"ORTH": "-", "OP": "?"}, {"SHAPE": "dddd"}]}
            ]
# adiciona os padrões na regra
ruler.add_patterns(patterns)

# cria um doc
doc = nlp(text)

# pega as entidades do doc
for ent in doc.ents:
    print (ent.text, ent.label_)
```
