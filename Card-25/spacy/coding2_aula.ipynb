```python
import spacy
import numpy as np
```

```python
# carregando o pipeline do ingles
nlp = spacy.load('en_core_web_md')
```

```python
# pegando o texto do arquivo
with open('files/wiki_us.txt', 'r') as f:
    text = f.read()
```

```python
# criando o doc
doc = nlp(text)
# pegando a primeira sentençá
sentence1 = list(doc.sents)[0]
print(sentence1)
```

```python
# palavra base
your_word = "country"

# filtra o vetor que contem as 10 palavras mais similares a palavra base
ms = nlp.vocab.vectors.most_similar(
    np.asarray([nlp.vocab.vectors[nlp.vocab.strings[your_word]]]), n=10)

# seleciona as palavras na lista e mostra
words = [nlp.vocab.strings[w] for w in ms[0][0]]
print(words)
```

```python
# criando dois novos docs
doc1 = nlp('I like salty fries and hamburgers.')
doc2 = nlp('Fast food tastes very good.')
```

```python
# vendo a similaridade entre os 2 docs
print(doc1, '<->', doc2, doc1.similarity(doc2))
```

```python
# comparando com um outro novo doc
doc3 = nlp('The Empire State Building is in New York')
print(doc1, '<->', doc3, doc1.similarity(doc3))
```

```python
# vendo o valor de similaridade para frases muito parecidas
doc4 = nlp('I enjoy oranges.')
doc5 = nlp('I enjoy apples.')

print(doc4, '<->', doc5, doc4.similarity(doc5))
```

```python
# testando com outra frase semelhante, mas com outra comida
doc6 = nlp('I enjoy burgers.')
print(doc4, '<->', doc6, doc4.similarity(doc6))
```

```python
# vendo a similaridade entre
french_fries = doc1[2:4]
burgers = doc1[5]
print(french_fries, '<->', burgers, french_fries.similarity(burgers))
```
