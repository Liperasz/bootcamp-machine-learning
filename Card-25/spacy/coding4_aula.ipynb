```python
import spacy
from spacy.matcher import Matcher
```

```python
# carregando o pipeline
nlp = spacy.load('en_core_web_md')
```

```python
# instanciando o matcher
matcher = Matcher(nlp.vocab)
```

```python
# criando um padrão
pattern = [
    {'LIKE_EMAIL': True}
]
# adicionando no matcher
matcher.add('EMAIL_ADDRESS', [pattern])
```

```python
# criando um doc
doc = nlp('This is an email address: cavalo123@utfpr.com')
# vendo as correspondencias do doc com o padrao estabelecido
matches = matcher(doc)
print(matches)
```

```python
# vendo o que o numero representa em texto
print(nlp.vocab[matches[0][0]].text)
```

```python
# abrindo o arquivo da wiki
with open('files/wiki_mlk.txt', 'r') as f:
    text = f.read()

print(text)
```

```python
# carregando o pipeline do ingles
nlp = spacy.load('en_core_web_md')
```

```python
# criando um padrão para reconher nomes proprios
pattern = [
    {'POS': 'PROPN', 'OP': '+'},
    {'POS': 'VERB'}
]
# adicionando o padrao no matcher, selecionando sempre a maior correspondencia
matcher.add('PROPER_NOUN', [pattern], greedy='LONGEST')
# criando o doc do texto
doc = nlp(text)

# vendo as correspondencias
matches = matcher(doc)
# ordenando
matches.sort(key = lambda x: x[1])
print(len(matches))
for match in matches[:10]:
    print(match, doc[match[1]:match[2]])
```

```python
# carregando o texto do arquivo json
import json
with open('files/alice.json', 'r') as f:
    data = json.load(f)
```

```python
# pegando o primeiro parágrafo do texto
text = data[0][2][0]
print(text)
```

```python
# substituindo/corringindo simbolos do texto
text = text.replace("`", "'")
print(text)
```

```python
# instanciando o matcher
matcher = Matcher(nlp.vocab)

# lista de lemmas
speak_lemmas = ['think', 'say']

# criando um padrao para reconhecer
pattern = [
    # reconhecer texto entre aspas
    {'ORTH': "'"},
    {'IS_ALPHA': True, 'OP': '+'},
    {'IS_PUNCT': True, 'OP': '*'},
    {'ORTH': "'"},
    # deve haver um verbo que está na lista de lemmas
    {'POS': 'VERB', 'LEMMA': {'IN': speak_lemmas}},
    # procurando quem que falou
    {'POS': 'PROPN','OP': '+'},
    # reconhecer como fez
    {'ORTH': "'"},
    {'IS_ALPHA': True, 'OP': '+'},
    {'IS_PUNCT': True, 'OP': '*'},
    {'ORTH': "'"}
]
# adicionando o padrao no matcher, selecionando sempre a maior correspondencia
matcher.add('PROPER_NOUN', [pattern], greedy='LONGEST')
# criando o doc do texto
doc = nlp(text)

# vendo as correspondencias
matches = matcher(doc)
# ordenando
matches.sort(key = lambda x: x[1])
print(len(matches))
for match in matches[:10]:
    print(match, doc[match[1]:match[2]])
```

```python
for text in data[0][2]:
    # criando o doc do texto
    text = text.replace("`", "'")
    doc = nlp(text)
    # vendo as correspondencias
    matches = matcher(doc)
    # quantidade de matches
    print(len(matches))
    # ordenando
    matches.sort(key = lambda x: x[1])
    for match in matches[:10]:
        print(match, doc[match[1]:match[2]])
```
