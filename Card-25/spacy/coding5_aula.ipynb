```python
import spacy
```

```python
# carregando o pipeline em ingles
nlp = spacy.load('en_core_web_sm')
# criando um doc
doc = nlp("Britain is a place. Mary is a doctor.")
```

```python
# vendo as entidades
for ent in doc.ents:
    print(ent.text, ent.label_)
```

```python
from spacy.language import Language
```

```python
# registrando o componente e a função dele
@Language.component('remove_gpe')
# função que remove a entidade GPE
def remove_gpe(doc):
    original_ents = list(doc.ents)
    for ent in doc.ents:
        if ent.label_ == 'GPE':
            original_ents.remove(ent)
    doc.ents = original_ents
    return doc
```

```python
# adiciona o componente no pipeline
nlp.add_pipe('remove_gpe')
```

```python
# criando um doc
doc = nlp('Britain is a place. Mary is a doctor.')
# vendo as entidades do doc
for ent in doc.ents:
    print(ent.text, ent.label_)
```
