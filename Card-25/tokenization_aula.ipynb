```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
# sentenças para aplicar no tokenizador
sentences = [
    'I love my dog',
    'I love my cat',
    'You love my dog!',
    'Do you think my dog is amazing?'
]
```

```python
# instanciando o tokenizador
# num words é a quantidade max de palavras no vocabulario
# oov é um token que vai ser utilizado para todas as palavras que nao estiverem no vocabulario
tokenizer = Tokenizer(num_words=100, oov_token='<00V>')
# faz o processo
tokenizer.fit_on_texts(sentences)
# index aplicados para cada palavra
word_index = tokenizer.word_index
# transforma as palavras nas sequencias de numeros usando o mapeamento
sequences = tokenizer.texts_to_sequences(sentences)
# aplicando o preenchimento nas sequencias
padded = pad_sequences(sequences, padding='post', truncating='post', maxlen=5)

print(word_index)
print(sequences)
print(padded)
```

```python
# sentenças de teste com palavras nao conhecidas
test_data = [
    'i really love my dog',
    'my dog loves my manatee',
]

# aplicando a tokenização
test_seq = tokenizer.texts_to_sequences(test_data)
print(test_seq)
```
