```python
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import InputLayer, Dense, Flatten, Conv2D, MaxPooling2D
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras import utils as np_utils
```

```python
# carregando o dataset
(X_train, y_train), (X_test, y_test) = mnist.load_data()
```

```python
# formatando
X_train = X_train.reshape(X_train.shape[0], 28, 28, 1)
X_test = X_test.reshape(X_test.shape[0], 28, 28, 1)
# transformando o tipo para float
X_train = X_train.astype('float32')
X_test = X_test.astype('float32')
# normalizando
X_train /= 255
X_test /= 255
# transformando em categorias
y_train = np_utils.to_categorical(y_train, 10)
y_test = np_utils.to_categorical(y_test, 10)
```

```python
# criando a rede neural
neural_network = Sequential()
neural_network.add(InputLayer(shape=(28,28,1)))
neural_network.add(Conv2D(32, (3,3), activation='relu'))
neural_network.add(MaxPooling2D(pool_size=(2,2)))
neural_network.add(Flatten())
neural_network.add(Dense(units=128, activation='relu'))
neural_network.add(Dense(units=10, activation='softmax'))
neural_network.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
# parametros para criar as novas imagens
generator_train = ImageDataGenerator(rotation_range=7, horizontal_flip=True, shear_range=0.2, height_shift_range=0.07, zoom_range=0.2)
```

```python
# imagens de teste não precisam de extras
generator_test = ImageDataGenerator()
```

```python
# criando a base de treinamento
base_train = generator_train.flow(X_train, y_train, batch_size=128)
```

```python
# base de teste
base_test = generator_test.flow(X_test, y_test, batch_size=128)
```

```python
# treinando o modelo
neural_network.fit(base_train, epochs=5, validation_data=base_test)
```
