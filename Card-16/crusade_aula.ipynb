```python
from keras.datasets import mnist
from keras.models import Sequential
from keras.layers import InputLayer, Dense, Flatten, Conv2D, MaxPooling2D
from keras import utils as np_utils
from sklearn.model_selection import StratifiedKFold
```

```python
# pré processamento

# baixando o dataset
(X, y), (X_test, y_test) = mnist.load_data()
# alterando para o formato com 1 camada de profundidade
X = X.reshape(X.shape[0], 28, 28, 1)
# convertendo para float
X = X.astype('float32')
# normalizando
X /= 255
# categorizando
y = np_utils.to_categorical(y, 10)
```

```python
# reprodutibilidade
seed = 5
np.random.seed(seed) 
```

```python
# definindo os parametros de separação
kfold = StratifiedKFold(n_splits=5, shuffle=True, random_state=seed)
```

```python
# mostrando os indices dos valores que vão ser utilizados para treinar e testar em cada uma das 5 separações
results = []
for index_train, index_test in kfold.split(X, np.zeros(shape=(y.shape[0],1))):
    print('Index Train: ', index_train, 'Index Test: ', index_test)

    # estrutura da rede neural
    neural_network = Sequential()
    neural_network.add(InputLayer(shape=(28,28,1)))
    neural_network.add(Conv2D(32, (3,3), activation='relu'))
    neural_network.add(MaxPooling2D(pool_size=(2,2)))
    neural_network.add(Flatten())
    neural_network.add(Dense(units=128, activation='relu'))
    neural_network.add(Dense(units=10, activation='softmax'))
    # compilação
    neural_network.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
    # treinamento
    neural_network.fit(X[index_train], y[index_train], batch_size=128, epochs=5)
    # precisão do modelo
    accuracy = neural_network.evaluate(X[index_test], y[index_test])
    # adicionando a lista dos resultados
    results.append(accuracy[1])
```

```python
# mostrando os resultados
results
```

```python
# média de precisão
np.array(results).mean()
```
