```python
import numpy as np
import matplotlib.pyplot as plt

# Velocidade de carregamento da página, com média de 3 segundos e desvio padrão de 1 segundo
pageSpeeds = np.random.normal(3.0, 1.0, 1000)
# Quantidade de compras, inverso a velocidade de carregamento da página
purchaseAmount = 100 - (pageSpeeds + np.random.normal(0, 0.1, 1000)) * 3

# Plotar os dados
plt.scatter(pageSpeeds, purchaseAmount)
```

```python
from scipy import stats

# Regressão linear]
# alope é o coeficiente angular
# intercept é o ponto onde a linha cruza o eixo y
# r_value é o coeficiente de correlação
# p_value é o valor p que indica a significância estatística
# std_err é o erro padrão da estimativa
alope, intercept, r_value, p_value, std_err = stats.linregress(pageSpeeds, purchaseAmount)
```

```python
# coeficiente de determinação
r_value ** 2
```

```python
# função para prever o valor de compra com base na velocidade da página
def predict(x):
    return alope * x + intercept

# criando a linha de previsão ao prever os valores utilizando as velocidades da página
fitLine = predict(pageSpeeds)

# plotando os dados
plt.scatter(pageSpeeds, purchaseAmount)
plt.plot(pageSpeeds, fitLine, c='red')
plt.show()
```

## Activity

Try increasing the random variation in the test data, and see what effect it has on the r-squared error value

```python
# aumentei o desvio padrão para 10
pageSpeeds = np.random.normal(3.0, 10.0, 1000)
# aumentei o desvio padrão para 5
purchaseAmount = 100 - (pageSpeeds + np.random.normal(0, 5, 1000)) * 3

# Plotar os dados
plt.scatter(pageSpeeds, purchaseAmount)
```

```python
# Regressão linear novamente
alope, intercept, r_value, p_value, std_err = stats.linregress(pageSpeeds, purchaseAmount)
```

```python
# função para prever o valor de compra com base na velocidade da página
def predict(x):
    return alope * x + intercept

fitLine = predict(pageSpeeds)

# plotando os dados
plt.scatter(pageSpeeds, purchaseAmount)
plt.plot(pageSpeeds, fitLine, c='red')
plt.show()
```

```python
# vendo o coeficiente de determinação novamente
r_value ** 2
```

Com o aumento do desvio padrão ao fornecer os dados, o coeficiente de determinação ficou menor, o que indica que esse modelo de previsão explica cerca de 80% da variação nos valores observados.
