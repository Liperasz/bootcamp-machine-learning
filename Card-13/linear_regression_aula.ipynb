```python
import numpy as np
import matplotlib.pyplot as plt

# Velocidade de carregamento da página, com média de 3 segundos e desvio padrão de 1 segundo
pageSpeeds = np.random.normal(3.0, 1.0, 1000)
# Quantidade de compras, inverso a velocidade de carregamento da página
purchaseAmount = 100 - (pageSpeeds + np.random.normal(0, 0.1, 1000)) * 3

# Plotar os dados
plt.scatter(pageSpeeds, purchaseAmount)
```
