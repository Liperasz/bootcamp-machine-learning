```python
# importando bibliotecas
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score
```

```python
# semente para reprodutibilidade
np.random.seed(2)

# gerando dados aleatṕorios
pageSpeeds = np.random.normal(3.0, 1.0, 100)
purchaseAmount = np.random.normal(50.0, 10.0, 100) / pageSpeeds

# criando scatter plot
plt.scatter(pageSpeeds, purchaseAmount)
```

```python
# separando os dados em treino e teste
trainX = pageSpeeds[:80]
testX = pageSpeeds[80:]

trainY = purchaseAmount[:80]
testY = purchaseAmount[80:]

# plotando os dados de treino e teste

plt.scatter(trainX, trainY)
```

```python
# plotando os dados de teste
plt.scatter(testX, testY)
```

```python
# transformando em array numpy
x = np.array(trainX)
y = np.array(trainY)

# ajustando o treinamento em um polinômio de grau 4
p4 = np.poly1d(np.polyfit(x, y, 4))
```

```python
# linspace para plotagem da curva ajustada
xp = np.linspace(0, 7, 100)

# axes é para configurar os eixos x e y
axes = plt.axes()
# definindo limites dos eixos x e y
axes.set_xlim([0, 7])
axes.set_ylim([0, 200])

# plotando os dados e a a linha traçada
plt.scatter(x, y) 
plt.plot(xp, p4(xp), c='r')
plt.show()
```

Fazendo o mesmo mas agora com os dados para teste

```python
# transformando em array numpy
testx = np.array(testX)
testy = np.array(testY)

# configurando os eixos x e y
axes = plt.axes()
axes.set_xlim([0, 7])
axes.set_ylim([0, 200])

# plotando os dados de teste e a linha traçada
plt.scatter(testx, testy)
plt.plot(xp, p4(xp), c='r')
plt.show()
```

```python
# verificando o r squared nos dados de teste
r2 = r2_score(testy, p4(testx))
r2
```

```python
# verificando o r squared nos dados de treino
r2 = r2_score(np.array(trainY), p4(np.array(trainX)))
r2
```
