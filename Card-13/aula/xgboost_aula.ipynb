```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import xgboost as xgb
```

```python
# carregando o dataset iris
iris = load_iris()

# numero de amostras e features
numSamples, numFeatures = iris.data.shape
print(numSamples)
print(numFeatures)
print(list(iris.target_names))
```

```python
# separando os dados de treino e teste
X_train, X_test, y_train, y_test = train_test_split(iris.data, iris.target, test_size=0.2, random_state=0)
```

```python
# formatando os dados para o xgboost
train = xgb.DMatrix(X_train, label=y_train)
test = xgb.DMatrix(X_test, label=y_test)
```

```python
# definindo os parametros do modelo
param = {
    'max_depth': 4, # profundidade da arvore
    'eta': 0.3, # taxa de aprendizado
    'objective': 'multi:softmax', # tipo de problema
    'num_class': 3 # numero de classes
}

epochs = 10 # numero de iteracoes
```

```python
# treinando o modelo
model = xgb.train(param, train, epochs)
```

```python
# fazendo as previsões de teste
predictions = model.predict(test)
predictions
```

```python
# fazendo a avaliação do modelo
accuracy_score(y_test, predictions)
```
