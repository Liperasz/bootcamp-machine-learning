```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import scale
```

```python
# criando falsas idades e rendas agrupadas em K clusters para n clientes
def createClusteredData(N, K):

    # reproduzibilidade
    np.random.seed(10)
    # número de pontos por cluster
    pointsPerCluster = float(N) / K
    # lista de pontos
    X = []

    # criando os clusters
    for i in range(K):
        
        # centróides aleatórios
        incomeCentroid = np.random.uniform(20000.0, 200000.0)
        # Idade entre 20 e 70
        ageCentroid = np.random.uniform(20.0, 70.0)

        # criando os pontos em torno dos centróides
        for j in range(int(pointsPerCluster)):
            X.append([np.random.normal(incomeCentroid, 10000.0),
                      np.random.normal(ageCentroid, 2.0)])
        
    # convertendo para array numpy
    X = np.array(X)
    return X
```

```python
# criando dados falsos
data = createClusteredData(100, 5)

# criando o modelo KMeans
model = KMeans(n_clusters=5)

# treinando o modelo com os dados escalonados
model = model.fit(scale(data))

# mostrando a que cluster cada ponto pertence
model.labels_
```

```python
# usando o matplot agora
plt.figure(figsize=(8, 6))
plt.scatter(data[:, 0], data[:, 1], c=model.labels_.astype(float))
plt.show()
```
