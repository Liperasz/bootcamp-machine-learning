```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
# frases para teste
sentences = [
    'I like eggs and ham',
    'I love chocolate and bunnies',
    'I hate onions'
]
```

```python
# numero maximo de palavras no volacabulario
MAX_VOCAB_SIZE = 20000
# instanciando o tokenizer
tokenizer = Tokenizer(num_words=MAX_VOCAB_SIZE)
# adicionando as sentenças
tokenizer.fit_on_texts(sentences)
# transformando em inteiros
sequences = tokenizer.texts_to_sequences(sentences)
sequences
```

```python
# vendo o mapeamento
tokenizer.word_index
```

```python
# preenchendo com padding utilizando as configurações padrões
data = pad_sequences(sequences)
data
```

```python
# Definindo o tamanho máximo de valor por "frase"
MAX_SEQUENCE_LENGHT = 5
data = pad_sequences(sequences, maxlen=MAX_SEQUENCE_LENGHT)
data
```

```python
# Preenchendo os valores depois inves de antes
data = pad_sequences(sequences, maxlen=MAX_SEQUENCE_LENGHT, padding='post')
data
```

```python
# definindo o tamanho maximo como um tamanho maior do que o atual existente
data = pad_sequences(sequences, maxlen=6)
data
```

```python
# truncation com os valores (dimunir o tamanho para menos do maior)
data = pad_sequences(sequences, maxlen=4)
data
```

```python
# tdiminuindo o valor depois inves do antes
data = pad_sequences(sequences, maxlen=4, truncating='post')
data
```
