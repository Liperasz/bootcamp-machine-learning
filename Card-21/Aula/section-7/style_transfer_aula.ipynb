```python
from torch_snippets import *
from torchvision import transforms as T
from torch.nn import functional as F
from torchvision.models import vgg19
device = 'cuda' if torch.cuda.is_available() else 'cpu'
```

```python
# pré processamento para normalizar os dados
preprocess = T.Compose([
    T.ToTensor(),
    T.Normalize(mean=[0.485,0.456,0.406], std=[0.229,0.224,0.225]),
    T.Lambda(lambda x: x.mul_(255))
])

# pós processamento para voltar ao normal
postprocess = T.Compose([
    T.Lambda(lambda x: x.mul_(1./255)),
    T.Normalize(mean=[-0.485/0.229, -0.456/0.224,-0.406/0.225], std=[1/0.229,1/0.224,1/0.255]),
])
```

```python
# calcula a matrix para o tensor
class GramMatrix(nn.Module):
    def forward(self,input):
        b,c,h,w = input.size()
        feat = input.view(b,c,h*w)
        G = feat@feat.transpose(1,2)
        G.div_(h*w)
        return G

# calcula o MSE entre os inputs na matrix e os targets
class GramMSELoss(nn.Module):
    def forward(self,input,target):
        out = F.mse_loss(GramMatrix()(input),target)
        return(out)

# vgg19 modificado para extrair mapas de características em camadas específicas
class vgg19_modified(nn.Module):
    def __init__(self):
        super().__init__()
        features = list(vgg19(pretrained=True).features)
        self.features = nn.ModuleList(features).eval()

    def forward(self, x, layers=[]):
        order = np.argsort(layers)
        _results, results = [], []

        for ix, model in enumerate(self.features):
            x = model(x)
            if ix in layers:
                _results.append(x)

        for o in order:
            results.append(_results[o])

        return results if layers is not [] else x
```

```python
vgg = vgg19_modified().to(device)
```

```python
# carregando as imagens
imgs = [Image.open(path).resize((512,512)).convert('RGB') for path in ['Fall_2022_web-images_Picasso_32.jpg','how-to-draw-an-elephant-featured-image-1200.png']]
# pré processamento da imagem
style_image, content_image = [preprocess(img).to(device)[None] for img in imgs]
```

```python
# clonando o conteudo da imagem para otimização
opt_img = content_image.data.clone()
opt_img.requires_grad = True
```

```python
# camadas de cada características
style_layers = [0, 5, 10, 19, 28]
content_layers = [21]
loss_layers = style_layers + content_layers
```

```python
# função para calcular o erro para cada tipo de camada
loss_fns = [GramMSELoss()] * len(style_layers) + [nn.MSELoss()] * len(content_layers)
loss_fns = [loss_fn.to(device) for loss_fn in loss_fns]
```

```python
# definindo os pesos das camadas
style_weights = [1000/n**2 for n in [64,128,256,512,512]]
content_weights = [1]
weights = style_weights + content_weights
```

```python
# definindo os targets
style_targets = [GramMatrix()(A).detach() for A in vgg(style_image, style_layers)]
content_targets = [A.detach() for A in vgg(content_image, content_layers)]
targets = style_targets + content_targets
```

```python
# maximo de iterações
max_iters = 500
# definindo optimizador
optimizer = optim.LBFGS([opt_img])
log = Report(max_iters)
```

```python
iters = 0
while iters < max_iters:
    def closure():
        global iters
        iters += 1
        optimizer.zero_grad()
        out = vgg(opt_img, loss_layers)
        # perdas dos mapa de características
        layer_losses = [weights[a] * loss_fns[a](A, targets[a]) for a, A in enumerate(out)]
        # valor da perda da iteração
        loss = sum(layer_losses)
        loss.backward()
        log.record(pos=iters, loss=loss, end='\r')
        return loss
    # chamada para atualizar a variável de otimização
    optimizer.step(closure)
```

```python
# plotar a imagem e o erro
log.plot(log=True)
with torch.no_grad():
    out_img = postprocess(opt_img[0]).permute(1,2,0)
show(out_img)
```
