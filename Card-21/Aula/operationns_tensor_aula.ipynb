```python
# 2 'pilhas' cada uma com uma linha e 10 colunas
a = torch.ones(2, 1, 10)
a
```

```python
# juntando as duas pilhas em um objeto só, com duas linhas e 10 colunas agora
squeezed = a.squeeze(1)
a.shape, squeezed.shape
```

```python
# dois stacks com 2 linhas e 10 colunas cada
a = torch.ones(2,2,10)
print(f'a = {a}')
print(f'Formato de a = {a.shape}')
```

```python
# não muda nadaa
squeezed = a.squeeze(1)
print(f'squeezed = {squeezed}')
print(f'Formato de squeezed = {squeezed.shape}')
```

```python
# dois stacks com 3 linhas e uma coluna cada
a = torch.ones(2,3,1)
print(f'a = {a}')
print(f'Formato de a = {a.shape}')
```

```python
# aqui sim alterou para apenas 1 stack
squeezed = a.squeeze(2)
print(f'squeezed = {squeezed}')
print(f'Formato de squeezed = {squeezed.shape}')
```

```python
# 2 linhas e 3 colunas
a = torch.ones(2,3)
print(f'a = {a}')
print(f'Formato de a = {a.shape}')

# alterando para 2 stacks com 3 linhas e 1 coluna
squeezed = a.unsqueeze(2)
print(f'squeezed = {squeezed}')
print(f'Formato de squeezed = {squeezed.shape}')
```

```python
# 2 linhas 3 colunas
a = torch.ones(2,3)

# usando None para criar dimensões no eixo especificado
print(a[None].shape)
print(a[:,None].shape)
print(a[:,:,None].shape)
```
