```python
# 2 'pilhas' cada uma com uma linha e 10 colunas
a = torch.ones(2, 1, 10)
a
```

```python
# juntando as duas pilhas em um objeto só, com duas linhas e 10 colunas agora
squeezed = a.squeeze(1)
a.shape, squeezed.shape
```

```python
# dois stacks com 2 linhas e 10 colunas cada
a = torch.ones(2,2,10)
print(f'a = {a}')
print(f'Formato de a = {a.shape}')
```

```python
# não muda nadaa
squeezed = a.squeeze(1)
print(f'squeezed = {squeezed}')
print(f'Formato de squeezed = {squeezed.shape}')
```

```python
# dois stacks com 3 linhas e uma coluna cada
a = torch.ones(2,3,1)
print(f'a = {a}')
print(f'Formato de a = {a.shape}')
```

```python
# aqui sim alterou para apenas 1 stack
squeezed = a.squeeze(2)
print(f'squeezed = {squeezed}')
print(f'Formato de squeezed = {squeezed.shape}')
```

```python
# 2 linhas e 3 colunas
a = torch.ones(2,3)
print(f'a = {a}')
print(f'Formato de a = {a.shape}')

# alterando para 2 stacks com 3 linhas e 1 coluna
squeezed = a.unsqueeze(2)
print(f'squeezed = {squeezed}')
print(f'Formato de squeezed = {squeezed.shape}')
```

```python
# 2 linhas 3 colunas
a = torch.ones(2,3)

# usando None para criar dimensões no eixo especificado
print(a[None].shape)
print(a[:,None].shape)
print(a[:,:,None].shape)
```

```python
# matriz 2, 4
x = torch.tensor([[1, 2, 3, 4], [5, 6, 7, 8]])
# matriz 4,3
y = torch.tensor([[1, 2, 3],
                  [2, 3, 4],
                  [4, 5, 6],
                  [7, 8, 9]])

# multiplicando as 2 matrizes
torch.matmul(x, y)
```

```python
# 1 matriz com 4 linhas e 5 colunas 
x = torch.randn(1, 4, 5)
print(f'Matriz x = {x}')

# concatenando a matriz x com outra matriz x
z = torch.cat([x, x], axis=1)
print(f'Matriz z = {z}')
```

```python
# 3 matriz com 20 linhas e 10 colunas cada
x = torch.randn(3, 20, 10)
# permutando, ou seja, alteranndo as dimensões, invertendo elas
z = x.permute(2, 0, 1)
print(f'Dimensões alteradas: {x.shape}, {z.shape}')
```
