```python
import torch
import plotly.graph_objs as go
import plotly.offline as pyo
```

### Criando um objeto Tensor

```python
# array de uma dimensão
sample = torch.tensor([10, 11])
sample.shape
```

```python
# criando uma matriz (2x2)
x = torch.tensor([[10,11], [1,2]])
x.shape
```

```python
# criando uma matriz (2x1)
y = torch.tensor([[10], [11]])
y.shape
```

```python
# criando um tensor de 3 dimensões
a = torch.tensor([[[1,2], [3,4]], [[5,6], [7,8]]])
a.shape
```

```python
# criando uma lista para todos os valores x y e z do objeto do tensor
x_vals = []
y_vals = []
z_vals = []
for i in range(a.shape[0]):
    for j in range(a.shape[1]):
        for k in range(a.shape[2]):
            x_vals.append(i)
            y_vals.append(j)
            z_vals.append(k)

# criando um scatter plot com o Plotly 
trace = go.Scatter3d(
    x=x_vals,
    y=y_vals,
    z=z_vals,
    mode='markers',
    marker=dict(
        size=5,
        color=a.flatten(),
        colorscale='Viridis',
        opacity=0.8
    )
)

# criando o layout do plot
layout = go.Layout(
    margin=dict(l=0, r=0, b=0, t=0),
    scene=dict(
        xaxis=dict(title='Dimension 1'),
        yaxis=dict(title='Dimension 2'),
        zaxis=dict(title='Dimension 3')
    )
)

# combinando o scatter plot e o layout
fig = go.Figure(data=[trace], layout=layout)
pyo.plot(fig, filename='tensor.html')
```
