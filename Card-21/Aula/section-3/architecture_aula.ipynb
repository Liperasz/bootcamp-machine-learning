```python
import torch
import torch.nn as nn
from torchviz import make_dot
```

```python
# dados inventados para treinamento
x = [[1, 2], [3, 4], [5, 6], [7, 8]]
y = [[3], [7], [11], [15]]
```

```python
# convertendo em tensor
X = torch.tensor(x).float()
Y = torch.tensor(y).float()
```

```python
# criando a classe da rede neural
class NeuralNetwork(nn.Module):
    def __init__(self):
        super().__init__()
        # criando uma camada linear (2 dados de entrada e 8 de saida)
        self.layer1 = nn.Linear(2,8)
        # função de ativação relu
        self.activation = nn.ReLU()
        # 8 dados de entrada 1 de saída (resposta)
        self.layer2 = nn.Linear(8,1)

    # passagem de uma camada para outra
    def forward(self, x):
        x = self.layer1(x)
        x = self.activation(x)
        x = self.layer2(x)
        return x
```

```python
# instanciando o modelo
model = NeuralNetwork()

# criando uma imagem png que mostra visualmente a arquitetura do modelo
make_dot(model(X), params=dict(model.named_parameters())).render("dense_network", format="png")
```
