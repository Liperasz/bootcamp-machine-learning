```python
from torch.utils.data import Dataset, DataLoader
import torch
import torch.nn as nn
from torch.optim import SGD
```

```python
# dados inventados para treinamento
x = [[1, 2], [3, 4], [5, 6], [7, 8]]
y = [[3], [7], [11], [15]]
```

```python
# convertendo em tensor
X = torch.tensor(x).float()
Y = torch.tensor(y).float()
```

```python
# criando dataset
class MyDataset(Dataset):
    def __init__(self, x, y):
        self.x = torch.tensor(x).float()
        self.y = torch.tensor(y).float()

    def __len__(self):
        return len(self.x)
    
    def __getitem__(self, ix):
        return self.x[ix], self.y[ix]
```

```python
# carregando dataset
ds = MyDataset(x,y)
```

```python
# defininndo batch size e embaralhando
dl = DataLoader(ds, batch_size=2, shuffle=True)
```

```python
# vendo todos os dados
for x,y in dl:
    print(x,y)
```

```python
# criando a classe da rede neural
class NeuralNetwork(nn.Module):
    def __init__(self):
        super().__init__()
        # criando uma camada linear (2 dados de entrada e 8 de saida)
        self.layer1 = nn.Linear(2,8)
        # função de ativação relu
        self.activation = nn.ReLU()
        # 8 dados de entrada 1 de saída (resposta)
        self.layer2 = nn.Linear(8,1)

    # passagem de uma camada para outra
    def forward(self, x):
        x = self.layer1(x)
        x = self.activation(x)
        x = self.layer2(x)
        return x
```

```python
# instanciando o modelo
model = NeuralNetwork()

# definindo a função de loss
loss_func = nn.MSELoss()

# definindo optmizador
opt = SGD(model.parameters(), lr= 0.001)
```

```python
losses = []
# 50 épocas
for _ in range(50):
    for data in dl:

        # gradientes em zero antes de cada época
        opt.zero_grad()
        x1, y1 = data
        loss_value = loss_func(model(x1), y1)
        loss_value.backward()

        # atualiza os pesos
        opt.step()
        losses.append(loss_value.detach().numpy())
```
